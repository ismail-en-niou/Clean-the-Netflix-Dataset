```python
import pandas as pd
```

```python
df = pd.read_csv("netflix_titles.csv")
df.info()
df.head()
display(df.describe(include="all").T)
```

```python
# df.dropna(inplace=True)
df
```

```python
missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing.sort_values("Missing_Percentage", ascending=False))
```

```python
print(df.shape)
collumns_to_fill = [
    "director",
    "cast",
    "country",
    "rating",
    "listed_in",
    "description"
]
for col in collumns_to_fill:
    df[col] = df[col].fillna("Unknown")
df
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    errors="coerce"
)

print(df["date_added"].dtype)
```

```python
duration_parts = (
    df["duration"]
    .astype("string")
    .str.strip()
    .str.extract(r"^(\d+)\s*(min|Season|Seasons)$", flags=0)
)

df["duration_value"] = pd.to_numeric(
    duration_parts[0],
    errors="coerce"
).astype("Int64")

df["duration_unit"] = duration_parts[1].replace({
    "Season": "Seasons"
})


display(df[["title", "duration", "duration_value", "duration_unit"]].head(10))
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicated show IDs:", df["show_id"].duplicated().sum())
```

```python
# Check data types
df.info()

# Check missing values again
display(
    df.isnull().sum()
    .sort_values(ascending=False)
)

# Check date type
print("Date type:", df["date_added"].dtype)

# Check the duration columns
display(df[[
    "title",
    "type",
    "duration",
    "duration_value",
    "duration_unit"
]].head(10))
```

```python
df.to_csv(
    "netflix_titles_cleaned.csv",
    index=False,
    encoding="utf-8"
)

print("Cleaned dataset exported successfully.")
```
